# 🧪 AegisForge-AI: Agent Orchestrator Comprehensive Testing
This notebook tests the full Agentic Pipeline across different industrial scenarios to verify air-gap compliance and accurate routing.

```python
import os
import sys
import logging

# Silence the noisy HTTPX logs from LangChain/Ollama
logging.getLogger('httpx').setLevel(logging.WARNING)
logging.getLogger('httpcore').setLevel(logging.WARNING)

sys.path.append(os.path.abspath('..'))

# Run the main agents notebook to load the compiled graph (app) into memory
# Import sovereign pipeline graph directly from pipeline module
from agent_orchestrator.pipeline import app, AgentState

from langchain_core.messages import HumanMessage

def print_stream(app, input_state, config):
    print("\n🚀 Starting Pipeline Execution...")
    for event in app.stream(input_state, config=config):
        for node_name, value in event.items():
            print(f"\n{'='*50}")
            if node_name == 'supervisor':
                print("🤖 [Supervisor Agent] Invoked: Analyzing task and deciding next steps...")
            elif node_name == 'vision_agent':
                print("👁️ [Vision Agent] Invoked: Extracting data from scans/drawings...")
            elif node_name == 'coder_agent':
                print("⚙️ [Coder Agent] Invoked: Executing deterministic engineering math/scripts...")
            elif node_name == 'reasoning_agent':
                print("⚖️ [Reasoning Agent] Invoked: Auditing statutory compliance and CVC rules...")
            elif node_name == 'chief_reviewer':
                print("🛡️ [Chief Reviewer] Invoked: Reviewing results and generating formal output...")
            elif node_name == 'direct_answer':
                print("⚡ [Adaptive Router] Invoked: Fast-tracking direct answer for simple query...")
            else:
                print(f"🔄 [Node: {node_name}] Executing...")
            
            if 'messages' in value and value['messages']:
                msg = value['messages'][-1].content
                if len(msg) > 500:
                    print(f"\n📝 Output (Truncated):\n{msg[:500]}...")
                else:
                    print(f"\n📝 Output:\n{msg}")
    print(f"\n{'='*50}\n✅ Pipeline Execution Complete!\n")
```

## Test 1: Adaptive Complexity Router (Simple Chit-Chat)
Testing if the system bypasses the massive multi-agent loop for simple questions.

```python
config = {"configurable": {"thread_id": "test_adaptive_router_1"}}
input_state = {
    "messages": [HumanMessage(content="Hello, what are you and what do you do?")]
}
print_stream(app, input_state, config)
```
