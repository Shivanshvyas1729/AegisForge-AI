```python
import os
import logging
from dotenv import load_dotenv, find_dotenv
from openai import AsyncOpenAI
```

```python
load_dotenv(find_dotenv(), override=True)
```

```python
AICREDICTS_API_KEY = os.getenv("AICREDICTS_API_KEY")
AICREDICTS_BASE_API = os.getenv("AICREDICTS_BASE_API")
```

```python
def get_openai_client() -> AsyncOpenAI:
    """Create an AsyncOpenAI client routed to the configured API gateway."""
    api_key = os.getenv("AICREDICTS_API_KEY")
    base_url = os.getenv("AICREDICTS_BASE_API")
    return AsyncOpenAI(
        base_url=base_url,
        api_key=api_key or "missing-key",
    )
```

```python
client = get_openai_client()

# Note: 'await' is required when using AsyncOpenAI.
# Note: 'ai21/jamba-large-1.7' currently returns 404 (endpoint not active on provider),
# so 'openai/gpt-4o-mini' or 'meta-llama/llama-3.1-8b-instruct' are tested and working.
response = await client.chat.completions.create(
    model="openai/gpt-4o-mini",
    messages=[
        {"role": "user", "content": "hi sumit"}
    ],
)
```

```python
reply_text = response.choices[0].message.content
print(reply_text)
```
