# Deliverable Publisher Tool Test

This notebook exercises the local document publishing tool. The expected output is a `.docx` note, a `.pdf` note, and a 64-character SHA-256 integrity seal.

The output paths should look like:
- data/output/11-V-102_Approval_Note.docx
- data/output/11-V-102_Approval_Note.pdf

```python
from pathlib import Path
from tools.deliverable_publisher_tool import deliverable_publisher_tool

inspection_data = {
    'equipment_id': '11-V-102',
    'equipment_name': 'HP Separator Drum',
    'material': '2.25Cr-1Mo',
    'inspection_date': '2026-09-20',
    'location': 'PSU Refinery'
}

calculation_data = {
    'design_pressure_mpa': 14.5,
    'measured_thickness_mm': 138.2,
    't_req_mm': 138.5,
    'delta_mm': -0.3,
    'corrosion_rate_mm_yr': 0.75,
    'remaining_life_years': 2.4,
    'status': 'ALERT'
}

compliance_data = {
    'applicable_clause': 'CVC Circular 02/02/2004 Clause 4.2',
    'clause_summary': 'Single-source emergency approval route considered valid for safety-critical component replacement.',
    'procurement_route': 'Emergency procurement with engineering approval',
    'approval_status': 'Pending PSU approval'
}

executive_summary = (
    'The vessel remains within the minimum design envelope with a narrow thickness margin. '
    'The integrity decision requires engineering review and endorsement before work authorization.'
)

result = deliverable_publisher_tool(inspection_data, calculation_data, compliance_data, executive_summary)
print(result)

assert 'docx_path' in result
assert 'pdf_path' in result
assert 'sha256_hash' in result
assert result['docx_path'].endswith('11-V-102_Approval_Note.docx')
assert result['pdf_path'].endswith('11-V-102_Approval_Note.pdf')
assert len(result['sha256_hash']) == 64
assert Path('data/output/11-V-102_Approval_Note.docx').exists()
assert Path('data/output/11-V-102_Approval_Note.pdf').exists()
print('DELIVERABLE_PUBLISHER_TESTS_PASSED')
```
