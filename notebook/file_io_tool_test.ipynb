# File IO Tool Sandbox Test

This notebook validates a safe, project-scoped file read/write tool.
The expected behavior is:
- write inside the workspace is allowed
- read inside the workspace is allowed
- traversal like ../.. is blocked

```python
from tools.file_io_tool import file_io_tool

write_result = file_io_tool('write', 'sample_data/file_io_tool_demo.txt', 'Equipment ID: 11-V-102\nStatus: OK')
print(write_result)
assert write_result['status'] == 'SUCCESS'
```

```python
read_result = file_io_tool('read', 'sample_data/file_io_tool_demo.txt')
print(read_result)
assert read_result['status'] == 'SUCCESS'
assert 'Equipment ID: 11-V-102' in read_result['data']
```

```python
blocked_result = file_io_tool('read', '../outside.txt')
print(blocked_result)
assert blocked_result['status'] == 'ERROR'
assert 'sandbox' in blocked_result['message'].lower() or 'outside' in blocked_result['message'].lower() or 'traversal' in blocked_result['message'].lower()
print('FILE_IO_TOOL_TESTS_PASSED')
```
